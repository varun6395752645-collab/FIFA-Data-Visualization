### Histogram

```python
bins=[40,50,60,70,80,90,100]

plt.hist(fifa["Overall"],bins=bins,color="#0bb3ad")

plt.xticks(bins)

plt.xlabel("Skills Level")
plt.ylabel("Number of Players")
plt.title("Distribution of palyer skills in FIFA 2018")

plt.show()
```

### Piechart

```python
left=fifa.loc[fifa["Preferred Foot"]=="Left"].count()[0]
right=fifa.loc[fifa["Preferred Foot"]=="Right"].count()[0]
```

```python
labels=["Left","Right"]
color=["#08755f","#a8e3d7"]

plt.pie([left,right],labels=labels,colors=color,autopct="%.2f %%")

plt.title("Foot Preference of FIFA Players")
plt.show()
```

```python
fifa["Weight"]=[int(x.strip("lbs")) if type(x)==str else x for x in fifa["Weight"]]
```

```python
fifa["Weight"][0]
```

```python
light=fifa.loc[fifa["Weight"]<125].count()[0]
light_medium=fifa.loc[(fifa.Weight>=125)&(fifa.Weight<150)].count()[0]
medium=fifa[(fifa.Weight>=150)&(fifa.Weight<175)].count()[0]
medium_heavy=fifa[(fifa.Weight>=175)&(fifa.Weight<200)].count()[0]
heavy=fifa[fifa.Weight>=200].count()[0]
```

```python
weights=[light,light_medium,medium,medium_heavy,heavy]
labels=["under 125","125-150","150-175","175-200","over 200"]

plt.style.use("ggplot")

plt.pie(weights,labels=labels,autopct="%0.2f %%",explode=(0.4,.2,0,0,0.4))

plt.title("Weight Distribution of FIFA Players(in lbs)")

plt.show()
```

### Box And Whisker Plots

```python
fifa["Club"]
```

```python
barcelona=fifa.loc[fifa["Club"]=="FC Barcelona"]["Overall"]
madrid=fifa.loc[fifa["Club"]=="Real Madrid"]["Overall"]
revs=fifa.loc[fifa["Club"]=="New England Revolution"]["Overall"]
```

```python
plt.style.use("default")
plt.figure(figsize=(5,8))
labels=["FC Barcelona","Real Madrid","NE Revolution"]
plt.boxplot([barcelona,madrid,revs],labels=labels)
plt.title("Professional Soccer Team Comparison")
plt.ylabel("FIFA Ovarall rating")
plt.show()
```
